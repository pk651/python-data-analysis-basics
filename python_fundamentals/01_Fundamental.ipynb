# Object collections in Python

There are 3 main data structures in Python:
- Tuples: immutable sequence of values ​​defined by `( )`
- Lists: modular sequence of values ​​defined by `[ ]`
- Dictionaries: key-value indexed values ​​defined by `{ }`

**All these structures allow you to store any type of object.**

In Python, whatever the structure, we use `[ ]` to access an element of a structure

- In a tuple: `tup1[0]` allows you to access the first element

**Be careful in Python, the first index is always 0**

## Dictionaries

As we saw with lists and tuples, from the basic types (int, float, etc.) it is possible to develop new types that we call constructed types.

A new example of a constructed type is the **dictionary**.

The elements of a list or a tuple are ordered and we access an element thanks to its position using a number that we call the **index** of the element.

A **dictionary** in Python will also allow us to gather elements but these will be identified by a **key**. We can make the analogy with a French dictionary where we access a definition with a word.

Unlike lists which are delimited by square brackets, we use **braces** for dictionaries.



### Creating a Dictionary

We saw below that it was possible to create a dictionary with **curly braces** surrounding the elements. Another possible approach is to create an empty dictionary and add the elements as you go. It is possible to use curly braces or the **dict()** keyword

```python
number_of_tires = {} # ou dict()
number_of_tires["car"] = 4
number_of_tires["bike"] = 2
number_of_tires
```

### Slice

Another advantage of lists is the ability to select a portion of a list using an indexing pattern [m:n+1] to retrieve all elements, from the nth to the nth (from element m inclusive to element n+1 exclusive). We then say that we retrieve a slice of the list, for example:

- my_list[start:end:step] 

```python
contacts = [
    ("Charlie", "Leclerc", "0987654321"),
    ("Alice", "Dupont", "0612345678"),
    ("Bob", "Martin", "0123456789"),   
    ("Boby", "Martin", "0123456789"),   
    ("Boba", "Martin", "0123456789"),   
]
```

```python
contacts[1:3]
```

```python
contacts[-2:]
```

```python
tuple_contact = ('Alice', 'Dupont', '0612345678')
tuple_contact[:2]
```

```python
print(contacts[0:2])
print(contacts[0:3])
print(contacts[0:])
print(contacts[:])
print(contacts[1:])
print(contacts[1:-1])
print(contacts[::2])
```

## Functional Programming

Python allows the definition of anonymous functions as lambdas which can in turn be used as arguments to other functions.

``lambda arguments: expression
``

- arguments: Une liste d'arguments séparés par des virgules.
- expression: Une seule expression qui retourne la valeur de la fonction.


```python
# lambda

function_lambda = lambda x: x * x

function_lambda(10)
```

There are 2 predefined functions that allow you to work with functions on sequences:
- `map`: `map(function, sequence)` returns an iterator on the application of the function on each element of the sequence
- `filter`: `filter(function, sequence)` returns an iterator on each element of the sequence for which the function returns `True`.

```python
# map
example_list = list(zip(range(10), list("Python 3.7"), range(20, 10, -1)))

print(example_list)

list(map(lambda x: x[1], example_list))
```

### Style

When using a programming language, it is very important to follow the conventions and constructs specific to the language. This makes it easy to share code with the community, and also to easily understand code written by others.

In a nutshell, you should learn to write in a Pythonic style. There are many references to this on the Internet, the starting point being the **PEP 0008 — Style Guide for Python code (official document)** written by Guido van Rossum.

*Why is this important?*
- **Readability**: Well-formatted code is easier to read and understand, even for developers who didn't write it.
- **Maintainability**: Consistent code is easier to modify and update.
- **Collaboration**: Common style conventions make it easier to work in teams.
- **Adoption**: PEP 8 is widely adopted in the Python community, making it a de facto standard.

*Key points covered by PEP 8*
- **Indentation**: Use 4 spaces for each level of indentation.
- **Line length**: Limit lines of code to 79 characters.
- **Spaces around operators**: Always leave a space before and after operators.
- **Variable and function naming**: Use descriptive names in lowercase, separating words with underscores (snake_case).
- **Imports**: Group imports at the top of the file, separating standard imports, third-party imports, and local imports.
- **Comments**: Use clear and concise comments to explain the code.
- **Strings**: Use single or double quotes consistently.
