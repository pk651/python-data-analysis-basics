# Processing tabular data in Python

## Multidimensional array `ndarray`

In this paragraph we will just introduce the `ndarray` data structure and its basic functions.

The other operations will be seen through the `Series` and `DataFrame` objects:
- arithmetic
- logic
- vector and matrix calculations
- statistical calculations
- graphical representations

````python
import numpy as np

# by default ndarrays are vectors (dimension 1)
a = np.arange(48)
b= np.arange(2,50,3)
print("b=",b)
print("a=",a)
print(type(b), a.ndim, a.shape)
print(a.reshape(6,8))
````

````python
print(type(a))
````

````python
# number of dimensions of an ndarray
a.ndim
````

````python
# dimensions of an ndarray
a.shape
````

````python
# number of elements in an ndarray
a.size
````

````python
# type of elements of an ndarray
a.dtype
````

It is possible to transform an *ndarray* into an array of any dimension (as long as the total number of elements is compatible).

````python
# example transformation
# table with 8 rows and 6 columns

a = a.reshape(6, 8)
a[:, 3]
````

````python
a[:, 3]
````

````python
# access to the first row
# print("x=", a[5,:])
# print("y=",a[5])
print("z=",a[:-1])
````

````python
# access to the second column
a[:, 1]
````

````python
# access to a subarray
a[2::2, 1::3]
````

````python
a.sum(axis=1)
````

````python
# Calculate the sum of the elements of the array
sum = np.sum(a)
print("Sum of elements:", sum)

# Calculate the mean of the elements of the array
mean = np.mean(a)
print("Mean of elements:", mean)

# Calculate the standard deviation of the elements of the array
standard_dev = np.std(a)
print("Standard deviation of elements:", standard_dev)

# Find the maximum and minimum of the array
maximum = np.max(a)
minimum = np.min(a)
print("Maximum:", maximum)
print("Minimum:", minimum)


# Calculate the sum of each column
sum_columns = np.sum(a, axis=0)
print("Sum of columns:", sum_columns)

# Calculate the sum of each row
sum_rows = np.sum(a, axis=1)
print("Sum of rows:", sum_rows)
````

# Merge with Pandas

We want to join datasets using keys (common variables)

- `pd.merge()` allows to join two DataFrames, we use as options `on='key'`

- We can use as option `how=`, we can have:
- `left` (resp. `right`) in this case, we keep the dataset on the left (resp. right) and for the data on the right missing values ​​are added.
- `inner`, we only keep the common values ​​of the two datasets
- `outer`, we keep all the values ​​of the two datasets
- We can have different keys and do a join on the keys with `left_on='key1'` and `right_on='key2'`
- We can have several keys and do a join on the two keys `on=['key1','key2']`

For more details: https://pandas.pydata.org/pandas-docs/dev/reference/api/pandas.DataFrame.merge.html

````python
import pandas as pd

# Create two DataFrames
df1 = pd.DataFrame({'key': ['K0', 'K1', 'K2'],
'col1': ['A0', 'A1', 'A2'],
'col2': ['B0', 'B1', 'B2']})

df2 = pd.DataFrame({'key': ['K0', 'K1', 'K3'],
'col3': ['C0', 'C1', 'C3'],
'col4': ['D0', 'D1', 'D3']})

# Inner join
merged_inner = pd.merge(df1, df2, on='key')
print(merged_inner)

# Outer join to LEFT
merged_left = pd.merge(df1, df2, on='key', how='left')
print(merged_left)

# Right outer join
merged_right = pd.merge(df1, df2, on='key', how='right')
print(merged_right)

# Full outer join
merged_outer = pd.merge(df1, df2, on='key', how='outer')
print(merged_outer)
````
