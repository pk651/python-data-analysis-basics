### Building a class

- Object-oriented programming is a programming style that allows you to group in the same place the behavior (functions) and data (structures) that are made to go together
- The notion of object in Python concerns all Python structures
- The idea is to create your own objects

Creating an object is done in two steps:
- **Attributes**: description of the object
- **Functions/methods**: manipulate the data (attributes) of the object on which it is called.

The notion of class in python represents the description of the object
The second step is done only by allocating information to the object. The object obtained is an instance of the class


```python
class MyClass:
    def __init__(self,val1=0,val2=0):
        """ this is the constructor of the class which allows 
        to initialize the attributes of the class"""
        self.val1 = val1
        self.val2 = val2
    
    def method1(self,param1):
        self.val1 += param1
        print(self.val1)
```

This class allows you to create an object that has two attributes:
- `val1` initialized to `0`
- `val2` initialized to `0`

It also has a method allowing you to add a value `param1` to `val1`

```python
# we can create an object of this class using:
objet_maclasse = MyClass(val1=10, val2=20)
```

```python
# we can print val1 using:
print(objet_maclasse.val1)
```

```python
# the method method1 will add 10 to val1 and print the result
objet_maclasse.method1(10)
print(objet_maclasse.val1)
```

**Polygon class**

A polygon is composed of a list of Points. 
To manage the list, we create a Polygon Class. 

```python
class Polygon:
    def __init__(self):
        self.points = []

    def __str__(self):
        return("Poly{%s}"%', '.join([str(p) for p in self.points]))

    def add_point(self, point):
        if not point in self.points: # to be sure the points are different
            self.points.append(point)

    def add_points(self, points):
        for point in points:
            self.add_point(point)

    def add_polygone(self, polygone):
        self.add_points(polygone.points)

    def name(self):
        nb = len(self.points)
        if nb < 6:
            return ["vide", "point", "segment", "triangle", "quadrilatère", "pentagone"][nb]
        return "polygone"
```

Define a class `BankAccount()`, which allows you to instantiate objects such as `account1`, `account2`, etc.

The constructor of this class will initialize two instance attributes `name` and `balance`, with default values ​​`'A'` and `0`.

Three other methods are defined:
- `deposit(amount)` will add a certain amount to the balance
- `withdraw(amount)` will withdraw a certain amount from the balance

Complete the next cells with the instructions in comment

```python
class BankAccount:
    def __init__(self, name='A', balance=0):
        # initialize two instance attributes `name` and `balance`, with default values ​​`'A'` and `0`.
        self.name = name
        self.balance = balance
    
    def deposit(self, amount):
        # add a certain amount to the balance
        self.balance += amount
    
    def withdraw(self, amount):
        # withdraw a certain amount from the balance
        self.balance -= amount
    
    def print(self):
        # print the account balance and an alert message in case of a negative balance.
        print(f"{self.name} has {self.balance}€")
        if self.balance < 0 :
            print("/!\ WARNING /!\ your balance is negative")
```

```python
# Do some transactions (deposit, withdraw) 

account1.withdraw(100)
account3.deposit(500)
account2.withdraw(45)
account3.withdraw(1000)
```

## Inheritance

Inheritance is a fundamental concept in object-oriented programming that allows you to create new classes by reusing the attributes and methods of an existing class. In Python, inheritance is simple to implement.

```python
class Vehicle:
    def __init__(self, brand):
        self.brand = brand

    def start(self):
        print(f"The vehicle {self.brand} starts.")

    def stop(self):
        print(f"The vehicle {self.brand} stops.")

class Car(Vehicle):
    def open_door(self):
        print("The door opens.")

class Bicycle(Vehicle):
    def do_ride(self):
        print("Ride!!")
    
    def stop(self):
        print(f"The bicyle stops.")
```

```python
my_bike = Bicycle("Kona")
my_bike.start()
my_bike.stop()
```
